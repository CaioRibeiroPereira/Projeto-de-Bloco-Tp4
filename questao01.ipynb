```python
import time
import random


# CLASSE PROCESSO
class Processo:
    def __init__(self, pid, burst):
        self.pid = pid
        self.burst = burst
        self.tempo_restante = burst
        self.estado = "Nova"

    def __str__(self):
        return f"P{self.pid}"


# HEAP MÍNIMA MANUAL
class MinHeap:
    def __init__(self):
        self.heap = []

    def inserir(self, processo):
        self.heap.append(processo)
        self.subir(len(self.heap) - 1)

    def subir(self, indice):
        pai = (indice - 1) // 2

        if (
            indice > 0
            and self.heap[indice].tempo_restante
            < self.heap[pai].tempo_restante
        ):
            self.heap[indice], self.heap[pai] = (
                self.heap[pai],
                self.heap[indice],
            )

            self.subir(pai)

    def remover_min(self):

        if len(self.heap) == 0:
            return None

        if len(self.heap) == 1:
            return self.heap.pop()

        raiz = self.heap[0]

        self.heap[0] = self.heap.pop()

        self.descer(0)

        return raiz

    def descer(self, indice):

        menor = indice

        esquerda = 2 * indice + 1
        direita = 2 * indice + 2

        if (
            esquerda < len(self.heap)
            and self.heap[esquerda].tempo_restante
            < self.heap[menor].tempo_restante
        ):
            menor = esquerda

        if (
            direita < len(self.heap)
            and self.heap[direita].tempo_restante
            < self.heap[menor].tempo_restante
        ):
            menor = direita

        if menor != indice:
            self.heap[indice], self.heap[menor] = (
                self.heap[menor],
                self.heap[indice],
            )

            self.descer(menor)

    def vazia(self):
        return len(self.heap) == 0

    def mostrar(self):
        return [f"P{p.pid}({p.tempo_restante})" for p in self.heap]



# MOSTRAR ESTADOS
def mostrar_tabela(processos, pronta, suspensa):

    print("\n==============================================================")
    print(f"{'PID':<5} {'BURST':<10} {'RESTANTE':<12} {'ESTADO'}")
    print("===================================================================")

    for p in processos:
        print(
            f"{p.pid:<5} "
            f"{p.burst:<10} "
            f"{p.tempo_restante:<12} "
            f"{p.estado}"
        )

    print("==========================================================")

    print(f"Fila Pronta:   {pronta.mostrar()}")
    print(f"Fila Suspensa: {suspensa.mostrar()}")

    print("============================================================\n")



# CONFIGURAÇÃO
QUANTUM = 3

processos = []

burst_values = random.sample(range(9, 19), 10)

for i in range(1, 11):
    burst = burst_values[i - 1]

    processos.append(Processo(i, burst))


fila_pronta = MinHeap()
fila_suspensa = MinHeap()



# PROCESSOS CHEGANDO
print("\nPROCESSOS:\n")

for p in processos:

    p.estado = "Pronta"

    fila_pronta.inserir(p)

mostrar_tabela(processos, fila_pronta, fila_suspensa)



# ESCALONAMENTO
while not fila_pronta.vazia() or not fila_suspensa.vazia():

    # RETORNO DA SUSPENSA
    if not fila_suspensa.vazia():

        if random.random() < 0.4:

            processo_retorno = fila_suspensa.remover_min()

            processo_retorno.estado = "Pronta"

            fila_pronta.inserir(processo_retorno)

            print(
                f"Processo {processo_retorno.pid} "
                f"voltou da fila suspensa."
            )


    # CPU OCIOSA
    if fila_pronta.vazia():

        print("CPU esta ociosa, aguardando processos...")
        time.sleep(1)
        continue


    # PEGA PROCESSO
    processo = fila_pronta.remover_min()

    processo.estado = "Executando"

    print(f"\nCPU executando Processo {processo.pid}")

    mostrar_tabela(processos, fila_pronta, fila_suspensa)

    time.sleep(1)


    # EXECUTA QUANTUM
    processo.tempo_restante -= QUANTUM


    # TERMINOU
    if processo.tempo_restante <= 0:

        processo.tempo_restante = 0

        processo.estado = "Terminada"

        print(f"Processo {processo.pid} terminou.")

    else:

        # VAI PARA SUSPENSA
        if random.random() < 0.3:

            processo.estado = "Suspensa"

            fila_suspensa.inserir(processo)

            print(
                f"Processo {processo.pid} "
                f"foi para fila suspensa."
            )

        # VOLTA PARA PRONTA
        else:

            processo.estado = "Pronta"

            fila_pronta.inserir(processo)

            print(
                f"Quantum encerrado do "
                f"Processo {processo.pid}."
            )

    time.sleep(1)


# FINAL
print("\n******* TODOS OS PROCESSOS TERMINARAM ***********")

mostrar_tabela(processos, fila_pronta, fila_suspensa)
```

# Relatório 

## Objetivo

* Simular o funcionamento de um escalonador de processos.
* Utilizar uma fila de prontos, uma fila suspensa e uma CPU para executar os processos.
* Priorizar os processos com menor tempo restante utilizando uma Heap Mínima.

## Funcionamento

* Foram criados 10 processos com tempos de burst entre 9 e 18.
* Cada processo passou pelos estados Nova, Pronta, Executando, Suspensa e Terminada.
* O quantum utilizado foi igual a 3 unidades de tempo.
* A fila de prontos e a fila suspensa foram implementadas usando Heap Mínima.

## Resultados Obtidos

* A Heap Mínima organizou corretamente os processos durante toda a simulação.
* Diversos processos foram movidos para a fila suspensa e posteriormente retornaram para a fila de prontos.
* O tempo restante dos processos foi reduzido a cada quantum até sua finalização.
* Em alguns momentos a CPU ficou ociosa por não existir nenhum processo disponível na fila de prontos.
* Processos menores terminaram mais rapidamente, enquanto alguns processos maiores demoraram mais devido às suspensões.

## Por fim

* O escalonamento funcionou corretamente durante toda a execução.
* A Heap Mínima permitiu selecionar sempre o processo com menor tempo restante.
